# Dog Breed Identification Dataset Exploration

## Open-Source Datasets Considered

For the Dog Breed Identification project, we looked at several open-source image datasets that could be used to train and test a machine learning model. The datasets considered were:
1. Kaggle Dog Breed Identification Dataset
This dataset contains labelled images of 120 different dog breeds. The expected output variable is breed, which represents the breed associated with each dog image.
2. Stanford Dogs Dataset
This dataset contains images of dogs grouped into 120 different breeds. The expected output variable is the dog breed/class label.
3. Oxford-IIIT Pet Dataset
This dataset contains images of different cat and dog breeds. The expected output variable that would be relevant to our application is the breed label.

## Selected Dataset

We selected the Kaggle Dog Breed Identification Dataset for the project. It provides labelled dog images together with a labels.csv file containing an image id and its corresponding breed. This makes the dataset suitable for training a machine learning image classification model whose expected output is the predicted dog breed.
